```python
system_prompt=""""
You are an helpful assistant that analyzes the contents of the website,
and provides a short summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the response in code blocks. - respond just with markdown.
"""

user_prompt="""
Here are the contents of the website.
Provide a short summary of this website.
if it includes news or announcements, then summarize these too.
"""
```

```python
def messages_for(website):
    return [
        {"role":"system", "content": system_prompt},
        {"role":"user","content": user_prompt + website}
    ]
```

```python
messages_for(shivika)
```

```python
#for fun.. made gemini snarky......

messages=[{"role": "system", "content":"You are a snarky assistant"},
          {"role": "user", "content":"what is 2+2?"}]
response=gemini.chat.completions.create(model="gemini-2.5-flash-lite",messages=messages)
response.choices[0].message.content
```

```python

def summarize_website(url):
    print(f"Fetching website {url}...")
    website = fetch_website_contents(url)
    response = gemini.chat.completions.create(
        model="gemini-2.5-flash-lite",
        messages=messages_for(website)
    )
    return response.choices[0].message.content
```

```python
summarize_website("https://shivika304.github.io/myPortfolio/")
```

```python
def display_summary(url):
    summary=summarize_website(url)
    # summary=response.choices[0].message.content
    display(Markdown(summary))
```

```python
display_summary("https://kunalkaushal.tech")
```
